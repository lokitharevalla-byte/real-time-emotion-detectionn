```python
from transformers import pipeline
from google.colab import files
from PIL import Image
import matplotlib.pyplot as plt
```

```python
emotion_classifier = pipeline(
    "image-classification",
    model="dima806/facial_emotions_image_detection"
)

print("Model loaded successfully!")
```

```python
uploaded = files.upload()
```

```python
# Get uploaded filename
image_name = list(uploaded.keys())[0]

# Open image
image = Image.open(image_name).convert("RGB")

# Predict emotion
results = emotion_classifier(image)

# Display image
plt.figure(figsize=(7, 7))
plt.imshow(image)
plt.axis("off")
plt.show()

# Display predictions
print("\nEmotion Predictions:\n")

for result in results:
    emotion = result["label"]
    confidence = result["score"] * 100

    print(f"{emotion.upper()} : {confidence:.2f}%")

# Best prediction
best_emotion = results[0]["label"]
best_confidence = results[0]["score"] * 100

print("\n---------------------------")
print("PREDICTED EMOTION:", best_emotion.upper())
print("CONFIDENCE:", round(best_confidence, 2), "%")
print("---------------------------")
```
